```python
import pandas as pd
import plotly.express as px

# 1. Load cleaned dataset
df = pd.read_csv('../data/superstore_cleaned.csv')

# 2. Executive KPI Metrics
print("=== WEEK 4 EXECUTIVE KPI SUMMARY ===")
print(f"Total Sales: ${df['sales'].sum():,.2f}")
print(f"Total Profit: ${df['profit'].sum():,.2f}")
print(f"Profit Margin: {(df['profit'].sum() / df['sales'].sum()) * 100:.2f}%")
print(f"Average Discount: {df['discount'].mean() * 100:.2f}%\n")

# 3. Interactive Sub-Category Profitability Visual
fig_subcat = px.bar(
    df.groupby('sub_category')['profit'].sum().reset_index().sort_values(by='profit'),
    x='profit',
    y='sub_category',
    orientation='h',
    color='profit',
    color_continuous_scale='RdYlGn',
    title="Sub-Category Net Profitability Distribution"
)
fig_subcat.show()
```
